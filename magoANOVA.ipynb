```python
from scipy.stats import f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import pandas as pd
import numpy as np

df = pd.read_csv("/content/checkpoint_magos_dados.csv")

df["habilidade_geral"] = (
    df["precisao_pct"]
    + df["potencia_pct"]
    + df["controle_pct"]
) / 3

medias = (
    df.groupby("mago")["habilidade_geral"]
      .mean()
      .sort_values(ascending=False)
)

print("MÉDIA DE HABILIDADE:")
print(medias)

melhor_mago = medias.index[0]
melhor_media = medias.iloc[0]

print("\nMaior média:")
print(melhor_mago, melhor_media)

grupos = [grupo["habilidade_geral"].values for _, grupo in df.groupby("mago")]

f, p = f_oneway(*grupos)

print("F =", f)
print("p =", p)

if p < 0.05:
    print("Existe diferença estatisticamente significativa entre os magos.")
else:
    print("Não existe evidência de diferença significativa entre os magos.")

# Tukey HSD
tukey = pairwise_tukeyhsd(
    endog=df["habilidade_geral"],
    groups=df["mago"],
    alpha=0.05
)

print(tukey)
#anova, tukey, acho a diferenca e dps faco a media
```
